# First Derivative

```python
#we will use numpy to process the data
import numpy as np 

# to make the figures
import matplotlib.pyplot as plt 
```

## Normal spectrum

```python
# Load any spectrum you have around
pathSpectrum = r"G:\My Drive\Trabajo\Medidas experimentales\2016 - 02 - Mosquitos\All Spectra\AC-13D-BN-160715-160715.27.dpt"
with open(pathSpectrum, 'rb') as tmp:
    avmi = (line.replace(b'\t',b',') for line in tmp)
    spectrum = np.genfromtxt(avmi, delimiter=',')

# And split it in wavenumbers and absorbances
wavenumbers = np.transpose(spectrum)[0]
absorbances = np.transpose(spectrum)[1] 

# This is how it looks
plt.figure(figsize=(14,7))
plt.plot(wavenumbers,absorbances)
plt.xlim(wavenumbers[0], wavenumbers[-1])
          
# labels
plt.xlabel('Wavenumber')
plt.ylabel('Absorbance')

plt.show()
```

## First derivative

```python
d1Absorbances = np.diff(absorbances)
# Since the first derivative is obtained by subtracting a point from the one preceding it, its array has one less item. So to compensate for this in the wavenumber, we remove the last point. 
d1Wavenumbers = wavenumbers[:-1] 

# This is how it looks
plt.figure(figsize=(14,7))
plt.plot(d1Wavenumbers,d1Absorbances)
plt.xlim(d1Wavenumbers[0], d1Wavenumbers[-1])
          
# labels
plt.xlabel('Wavenumber')
plt.ylabel('Absorbance')

plt.show()
```

## Second derivative

Some people even use the second derivative

```python
d2Absorbances = np.diff(d1Absorbances)
# Since the first derivative is obtained by subtracting a point from the one preceding it, its array has one less item. So to compensate for this in the wavenumber, we remove the last point. 
d2Wavenumbers = d1Wavenumbers[:-1] 

# This is how it looks
plt.figure(figsize=(14,7))
plt.plot(d2Wavenumbers,d2Absorbances)
plt.xlim(d2Wavenumbers[0], d2Wavenumbers[-1])
          
# labels
plt.xlabel('Wavenumber')
plt.ylabel('Absorbance')

plt.show()
```
